```python
import tiktoken
from IPython.display import Markdown, display
```

```python
MODEL = "gpt-4o"
enc = tiktoken.encoding_for_model(MODEL)
```

```python
def token_count(text):
    return len(enc.encode(text))

broad_prompt   = "Tell me everything about machine learning"
focused_prompt = ("What's the difference between classification and regression? "
                  "Give one real-world example of each, under 100 words total.")
```

```python
print(f"Model: {MODEL}")

base_prompt_tokens = token_count(broad_prompt)
focused_prompt_tokens = token_count(focused_prompt)

print(f"Broad prompt:   {base_prompt_tokens} tokens")
print(f"Focused prompt: {focused_prompt_tokens} tokens")
```

```python
markdown_table_data = f"""
| Prompt | len(Plain Text) | len(Token) | Compression % |
| ---| --- | --- | --- |
| Base Prompt | {len(broad_prompt)} | {base_prompt_tokens} | {(base_prompt_tokens/len(broad_prompt))*100:.2f} |
| Focused Prompt | {len(focused_prompt)} | {focused_prompt_tokens} | {(focused_prompt_tokens/len(focused_prompt))*100:.2f} |
"""
display(Markdown(markdown_table_data))
```
